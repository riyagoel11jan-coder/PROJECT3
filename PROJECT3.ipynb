## <span style="color:purple"><b>About Exploratory Data Analysis (EDA)</b></span>

**Exploratory Data Analysis (EDA)** is a systematic process used to understand, summarize, and visualize data before advanced analysis or machine learning.

EDA helps in identifying **patterns, trends, relationships, and anomalies** present in the dataset. It plays an important role in understanding data quality and structure by detecting **missing values, duplicates, outliers, and inconsistencies**.

The main purpose of EDA is to convert **raw data into meaningful information** that supports reliable analysis and decision-making.

EDA generally consists of **Univariate, Bivariate, and Multivariate Analysis**, each focusing on different levels of data relationships and behavior.

Through statistical measures and graphical visualization, EDA improves understanding of the dataset and provides a strong foundation for **prediction, modeling, and business insights**.

Therefore, EDA is considered the **foundation of data science, analytics, and evidence-based decision-making**.

## <span style="color:purple"><b>Missing Value Analysis</b></span>

Missing value analysis is a critical stage of **Exploratory Data Analysis (EDA)** that focuses on identifying incomplete or unavailable information within the dataset. Missing values may arise due to data entry errors, system limitations, survey non-responses, or incomplete records.

The presence of missing data can significantly influence **statistical accuracy, visualization quality, and machine learning performance**. Therefore, it is essential to detect and quantify missing information before proceeding with advanced analysis.

This step helps evaluate **data completeness and reliability**, enabling informed preprocessing and cleaning decisions.

### <span style="color:lightblue"><b>Objectives of Missing Value Analysis</b></span>

- **Identify columns containing missing values**
- **Measure the quantity of missing observations**
- **Calculate missing value percentages**
- **Assess dataset quality and completeness**
- **Support data cleaning and preprocessing strategies**
- **Reduce analytical bias and improve reliability**

```python
missing_df = pd.DataFrame({

    "Missing":
    df.isnull().sum(),

    "Percentage":
    (df.isnull().sum() / len(df)) * 100
})

print("\nMISSING VALUES")
print(missing_df)
```

## <span style="color:lightpink"><b>Interpreting Missing Values</b></span>

The missing value summary helps determine the extent to which incomplete data may affect analysis.

General interpretation guidelines include:

- **0% Missing** → Complete and reliable data
- **1–5% Missing** → Minor missingness, usually manageable
- **5–20% Missing** → Moderate missingness requiring treatment
- **Above 20% Missing** → Significant missingness that may require advanced imputation or variable removal

Understanding missing value patterns helps ensure **accurate, reliable, and unbiased analytical outcomes**.

## <span style="color:purple"><b>Missing Value Treatment and Data Imputation</b></span>

After identifying missing values, the next stage involves **handling and imputing incomplete data** to improve dataset quality and analytical reliability. Missing data can negatively influence statistical accuracy, distort visualizations, and reduce the performance of predictive models if left untreated.

To address this challenge, the project applies **data imputation techniques**, which replace missing observations with statistically appropriate values rather than discarding entire records.

Different imputation strategies are applied based on the nature of variables to ensure meaningful and representative replacements.

This approach helps preserve information while minimizing bias and maintaining dataset integrity.

### <span style="color:lightblue"><b>Objectives of Missing Value Treatment</b></span>

- **Handle incomplete observations systematically**
- **Improve data quality and consistency**
- **Preserve valuable information**
- **Reduce analytical distortion**
- **Prepare data for statistical analysis and machine learning**
- **Maintain dataset reliability and interpretability**

```python
num_imputer = SimpleImputer(strategy='median')

cat_imputer = SimpleImputer(
    strategy='most_frequent'
)

if len(numerical_columns) > 0:

    df[numerical_columns] = num_imputer.fit_transform(
        df[numerical_columns]
    )

if len(categorical_columns) > 0:

    df[categorical_columns] = cat_imputer.fit_transform(
        df[categorical_columns])
```

🌊 Kernel Density Estimation (KDE) Analysis

</h2>

Kernel Density Estimation (KDE) is a statistical visualization technique used to estimate the probability density function of a numerical variable. Unlike histograms, KDE plots provide a smooth representation of the distribution and are less affected by bin selection. This analysis helps identify concentration areas, peaks, multiple modes, distribution symmetry, and overall data behaviour. KDE plots are particularly useful for understanding how observations are distributed across different value ranges and whether the variable follows a normal or skewed distribution pattern.

```python
# ============================================================
# KDE ANALYSIS
# ============================================================

for col in numerical_columns:

    data = df[col].dropna()

    mean_val = data.mean()
    median_val = data.median()
    mode_val = data.mode().iloc[0]
    std_val = data.std()
    var_val = data.var()
    min_val = data.min()
    max_val = data.max()
    skew_val = data.skew()
    kurt_val = data.kurtosis()

    cv_val = (
        std_val / mean_val * 100
        if mean_val != 0
        else 0
    )

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(14, 5),
        gridspec_kw={'width_ratios': [2, 1]}
    )

    sns.kdeplot(
        data,
        fill=True,
        color='royalblue',
        linewidth=2,
        ax=axes[0]
    )

    axes[0].set_title(
        f"KDE Plot - {col}",
        fontsize=14,
        fontweight='bold'
    )

    axes[0].set_xlabel(col)
    axes[0].set_ylabel("Density")

    stats_text = f"""
Count      : {data.count():,.0f}
Mean       : {mean_val:.2f}
Median     : {median_val:.2f}
Mode       : {mode_val:.2f}
Std Dev    : {std_val:.2f}
Variance   : {var_val:.2f}
Min        : {min_val:.2f}
Max        : {max_val:.2f}
Skewness   : {skew_val:.2f}
Kurtosis   : {kurt_val:.2f}
CV         : {cv_val:.2f}%
"""

    axes[1].axis("off")

    axes[1].text(
        0,
        0.5,
        stats_text,
        fontsize=11,
        bbox=dict(
            facecolor="#E3F2FD",
            edgecolor="royalblue",
            linewidth=2
        )
    )

    plt.tight_layout()
    plt.show()

    print("KEY INSIGHTS")
    print("=" * 70)

    print(f"• Variable analysed: {col}")

    print(f"• Total valid observations: {data.count():,}")

    print(f"• Mean value: {mean_val:.2f}")

    print(f"• Median value: {median_val:.2f}")

    print(f"• Mode value: {mode_val:.2f}")

    print(f"• Standard deviation: {std_val:.2f}")

    print(f"• Variance: {var_val:.2f}")

    print(f"• Minimum value: {min_val:.2f}")

    print(f"• Maximum value: {max_val:.2f}")

    print(f"• Range: {(max_val - min_val):.2f}")

    print(f"• Coefficient of variation: {cv_val:.2f}%")

    print(f"• Skewness: {skew_val:.2f}")

    print(f"• Kurtosis: {kurt_val:.2f}")

    if skew_val > 0.5:
        print("• Distribution pattern: Positively skewed / right-tailed.")

    elif skew_val < -0.5:
        print("• Distribution pattern: Negatively skewed / left-tailed.")

    else:
        print("• Distribution pattern: Approximately symmetric.")

    print()
```

📦 Box Plot Analysis

</h2>

A Box Plot is a statistical visualization that summarizes a numerical variable using the five-number summary: minimum, first quartile (Q1), median, third quartile (Q3), and maximum. It provides a compact representation of data spread, central tendency, and variability. Box plots are particularly useful for identifying outliers and understanding how observations are distributed around the median. This analysis helps detect unusual business behaviour, extreme observations, and variability among companies. It also serves as the foundation for IQR-based outlier detection.

```python
# ============================================================
# BOX PLOT ANALYSIS
# ============================================================

for col in numerical_columns:

    data = df[col].dropna()

    q1 = data.quantile(0.25)
    q2 = data.median()
    q3 = data.quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = data[
        (data < lower_bound) |
        (data > upper_bound)
    ]

    outlier_count = len(outliers)

    outlier_pct = (
        outlier_count / len(data)
    ) * 100

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(14,5),
        gridspec_kw={'width_ratios':[2,1]}
    )

    sns.boxplot(
        x=data,
        color='mediumorchid',
        ax=axes[0]
    )

    axes[0].set_title(
        f"Box Plot - {col}",
        fontsize=14,
        fontweight='bold'
    )

    stats_text = f"""
Q1 : {q1:.2f}

Median : {q2:.2f}

Q3 : {q3:.2f}

IQR : {iqr:.2f}

Lower Bound :
{lower_bound:.2f}

Upper Bound :
{upper_bound:.2f}

Outliers :
{outlier_count}

Outlier % :
{outlier_pct:.2f}%
"""

    axes[1].axis("off")

    axes[1].text(
        0,
        0.5,
        stats_text,
        fontsize=11,
        bbox=dict(
            facecolor="#F3E5F5",
            edgecolor="purple",
            linewidth=2
        )
    )

    plt.tight_layout()
    plt.show()

    print("KEY INSIGHTS")
    print("=" * 70)

    print(f"• Variable Analysed: {col}")

    print(f"• Q1: {q1:.2f}")

    print(f"• Median: {q2:.2f}")

    print(f"• Q3: {q3:.2f}")

    print(f"• IQR: {iqr:.2f}")

    print(f"• Lower Bound: {lower_bound:.2f}")

    print(f"• Upper Bound: {upper_bound:.2f}")

    print(f"• Outlier Count: {outlier_count}")

    print(f"• Outlier Percentage: {outlier_pct:.2f}%")

    print()
```

🚨 Outlier Detection Analysis

</h2>

Outlier detection is performed to identify observations that deviate significantly from the majority of the dataset. Outliers can arise due to exceptional business performance, unusual market conditions, data entry errors, or unique operational characteristics. In this analysis, two commonly used methods are applied: the Interquartile Range (IQR) method and the Z-Score method. The IQR method identifies observations lying beyond the normal quartile range, while the Z-Score method measures how many standard deviations an observation is from the mean. Using both techniques provides a comprehensive understanding of extreme values within the dataset.

```python
# ============================================================
# IQR + Z-SCORE OUTLIER ANALYSIS
# ============================================================

from scipy.stats import zscore

for col in numerical_columns:

    data = df[col].dropna()

    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    iqr_outliers = data[
        (data < lower) |
        (data > upper)
    ]

    z_scores = np.abs(zscore(data))

    z_outliers = data[z_scores > 3]

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(14,5)
    )

    # IQR
    sns.boxplot(
        x=data,
        color='crimson',
        ax=axes[0]
    )

    axes[0].set_title(
        f'IQR Outliers - {col}',
        fontweight='bold'
    )

    # Z Score
    axes[1].hist(
        z_scores,
        bins=25,
        color='teal',
        edgecolor='black'
    )

    axes[1].axvline(
        3,
        color='red',
        linestyle='--',
        linewidth=2
    )

    axes[1].set_title(
        f'Z-Score Distribution - {col}',
        fontweight='bold'
    )

    plt.tight_layout()
    plt.show()

    print("KEY INSIGHTS")
    print("="*70)

    print(f"• Variable: {col}")

    print(f"• IQR Outliers: {len(iqr_outliers)}")

    print(
        f"• IQR Outlier %: {(len(iqr_outliers)/len(data))*100:.2f}%"
    )

    print(f"• Z-Score Outliers: {len(z_outliers)}")

    print(
        f"• Z-Score Outlier %: {(len(z_outliers)/len(data))*100:.2f}%"
    )

    print(f"• Lower Bound: {lower:.2f}")

    print(f"• Upper Bound: {upper:.2f}")

    print()
```

• The Z-Score method identifies observations that are statistically distant from the mean.

🔗 Bivariate Analysis

</h1>

Bivariate Analysis is performed to study the relationship between two variables. Unlike univariate analysis, which focuses on a single variable, bivariate analysis helps identify associations, dependencies, trends, and patterns between variables. This stage is essential for understanding how business metrics influence one another and for identifying potential drivers of performance. The insights obtained from bivariate analysis support decision-making, forecasting, and predictive modelling.

```python
# ============================================================
# BIVARIATE STATISTICS
# ============================================================

corr_matrix = df[numerical_columns].corr()

upper_triangle = corr_matrix.where(
    np.triu(
        np.ones(corr_matrix.shape),
        k=1
    ).astype(bool)
)

corr_pairs = (
    upper_triangle
    .stack()
    .reset_index()
)

corr_pairs.columns = [
    "Variable 1",
    "Variable 2",
    "Correlation"
]

corr_pairs = corr_pairs.sort_values(
    "Correlation",
    ascending=False
)

corr_pairs.head(10)

strongest_positive = corr_pairs.iloc[0]

strongest_negative = corr_pairs.iloc[-1]

avg_corr = corr_pairs["Correlation"].mean()

strong_corr = (
    corr_pairs["Correlation"].abs() > 0.7
).sum()

moderate_corr = (
    (
        corr_pairs["Correlation"].abs() >= 0.3
    )
    &
    (
        corr_pairs["Correlation"].abs() <= 0.7
    )
).sum()

weak_corr = (
    corr_pairs["Correlation"].abs() < 0.3
).sum()

print("KEY INSIGHTS")
print("="*70)

print(
    f"• Strongest Positive Relationship:"
)

print(
    f"  {strongest_positive['Variable 1']} ↔ {strongest_positive['Variable 2']}"
)

print(
    f"  Correlation = {strongest_positive['Correlation']:.2f}"
)

print()

print(
    f"• Strongest Negative Relationship:"
)

print(
    f"  {strongest_negative['Variable 1']} ↔ {strongest_negative['Variable 2']}"
)

print(
    f"  Correlation = {strongest_negative['Correlation']:.2f}"
)

print()

print(
    f"• Average Correlation: {avg_corr:.2f}"
)

print(
    f"• Strong Relationships (|r| > 0.7): {strong_corr}"
)

print(
    f"• Moderate Relationships (0.3 ≤ |r| ≤ 0.7): {moderate_corr}"
)

print(
    f"• Weak Relationships (|r| < 0.3): {weak_corr}"
)
```

```python
# ============================================================
# LOOP-BASED CUMULATIVE BIVARIATE CORRELATION STATISTICS
# ============================================================

bivariate_stats = []

for i in range(len(numerical_columns)):

    for j in range(i + 1, len(numerical_columns)):

        col1 = numerical_columns[i]
        col2 = numerical_columns[j]

        corr_value = df[col1].corr(df[col2])

        if abs(corr_value) >= 0.7:
            strength = "Strong"

        elif abs(corr_value) >= 0.3:
            strength = "Moderate"

        else:
            strength = "Weak"

        bivariate_stats.append({
            "Variable 1": col1,
            "Variable 2": col2,
            "Correlation": corr_value,
            "Absolute Correlation": abs(corr_value),
            "Relationship Strength": strength
        })

bivariate_stats_df = pd.DataFrame(bivariate_stats)

bivariate_stats_df = bivariate_stats_df.sort_values(
    by="Absolute Correlation",
    ascending=False
)

display(bivariate_stats_df)
```

🧩 Multivariate Analysis

</h1>

Multivariate Analysis examines relationships among three or more variables simultaneously. Unlike univariate and bivariate analysis, which focus on one or two variables, multivariate techniques help uncover complex interactions, hidden patterns, clusters, and dependencies across multiple dimensions. This analysis provides a deeper understanding of business performance drivers and supports advanced analytics, segmentation, forecasting, and machine learning applications.

```python
# ============================================================
# CUMULATIVE MULTIVARIATE STATISTICS
# ============================================================

corr_matrix = df[numerical_columns].corr()

multivariate_stats = pd.DataFrame({

    "Statistic": [

        "Numerical Variables",

        "Total Variable Pairs",

        "Average Correlation",

        "Maximum Correlation",

        "Minimum Correlation",

        "Strong Relationships",

        "Moderate Relationships",

        "Weak Relationships"

    ],

    "Value": [

        len(numerical_columns),

        len(corr_matrix.stack()) - len(numerical_columns),

        round(
            corr_matrix.where(
                ~np.eye(
                    len(corr_matrix),
                    dtype=bool
                )
            ).stack().mean(),
            2
        ),

        round(
            corr_matrix.where(
                ~np.eye(
                    len(corr_matrix),
                    dtype=bool
                )
            ).stack().max(),
            2
        ),

        round(
            corr_matrix.where(
                ~np.eye(
                    len(corr_matrix),
                    dtype=bool
                )
            ).stack().min(),
            2
        ),

        (
            corr_matrix.where(
                ~np.eye(
                    len(corr_matrix),
                    dtype=bool
                )
            ).stack().abs() > 0.7
        ).sum(),

        (
            (
                corr_matrix.where(
                    ~np.eye(
                        len(corr_matrix),
                        dtype=bool
                    )
                ).stack().abs() >= 0.3
            )
            &
            (
                corr_matrix.where(
                    ~np.eye(
                        len(corr_matrix),
                        dtype=bool
                    )
                ).stack().abs() <= 0.7
            )
        ).sum(),

        (
            corr_matrix.where(
                ~np.eye(
                    len(corr_matrix),
                    dtype=bool
                )
            ).stack().abs() < 0.3
        ).sum()

    ]
})

display(multivariate_stats)

print("KEY INSIGHTS")
print("=" * 70)

print(
    f"• Numerical Variables Analysed: {len(numerical_columns)}"
)

print(
    f"• Total Relationships Evaluated: {multivariate_stats.iloc[1,1]}"
)

print(
    f"• Average Correlation Across Variables: {multivariate_stats.iloc[2,1]}"
)

print(
    f"• Strongest Relationship: {multivariate_stats.iloc[3,1]}"
)

print(
    f"• Weakest Relationship: {multivariate_stats.iloc[4,1]}"
)

print(
    f"• Strong Relationships Found: {multivariate_stats.iloc[5,1]}"
)

print(
    f"• Moderate Relationships Found: {multivariate_stats.iloc[6,1]}"
)

print(
    f"• Weak Relationships Found: {multivariate_stats.iloc[7,1]}"
)
```

🎯 Principal Component Analysis (PCA)

</h2>

Principal Component Analysis (PCA) is a dimensionality reduction technique that transforms multiple correlated variables into a smaller number of uncorrelated components called principal components. These components capture the maximum amount of variation present in the dataset. PCA helps simplify complex datasets, identify dominant patterns, reduce redundancy, and improve the efficiency of clustering and machine learning algorithms. It is particularly useful when dealing with multiple numerical variables that may contain overlapping information.

```python
# ============================================================
# PRINCIPAL COMPONENT ANALYSIS (PCA)
# ============================================================

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

pca_features = df[numerical_columns]

scaled_data = StandardScaler().fit_transform(
    pca_features
)

pca = PCA()

pca_data = pca.fit_transform(
    scaled_data
)

explained_variance = (
    pca.explained_variance_ratio_ * 100
)

cumulative_variance = (
    np.cumsum(explained_variance)
)
```

```python
# ============================================================
# PCA EXPLAINED VARIANCE
# ============================================================

plt.figure(figsize=(10,6))

bars = plt.bar(
    range(
        1,
        len(explained_variance)+1
    ),
    explained_variance,
    color=plt.cm.magma(
        np.linspace(
            0.25,
            0.90,
            len(explained_variance)
        )
    )
)

plt.plot(
    range(
        1,
        len(cumulative_variance)+1
    ),
    cumulative_variance,
    marker="o",
    linewidth=2.5,
    color="#355070"
)

plt.title(
    "PCA Explained Variance",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Principal Components",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Variance Explained (%)",
    fontsize=12,
    fontweight="bold"
)

for bar in bars:

    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{bar.get_height():.1f}%",
        ha="center",
        fontsize=9
    )

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()
plt.show()

pca_summary = pd.DataFrame({

    "Principal Component":
    [f"PC{i+1}" for i in range(len(explained_variance))],

    "Variance Explained (%)":
    explained_variance.round(2),

    "Cumulative Variance (%)":
    cumulative_variance.round(2)

})

display(pca_summary)

print("\nKEY INSIGHTS")
print("=" * 70)

print(
    f"• PC1 explains {explained_variance[0]:.2f}% of total variance."
)

print(
    f"• PC2 explains {explained_variance[1]:.2f}% of total variance."
)

print(
    f"• First 2 Components explain {cumulative_variance[1]:.2f}% of total variance."
)

print(
    f"• Components Required for 90% Variance: {(cumulative_variance < 90).sum() + 1}"
)

print(
    f"• Total Numerical Variables Analysed: {len(numerical_columns)}"
)
```

color:black;

</h2>

• PCA reveals how much information each principal component contributes to explaining the overall variability in the dataset.

• A large percentage of variance explained by the first few components indicates that the dataset can be simplified without losing substantial information.

• Components contributing very little variance may represent redundant or less informative dimensions.

🎯 Elbow Method

</h2>

The Elbow Method is used to determine the optimal number of clusters for K-Means clustering. It measures the Within-Cluster Sum of Squares (WCSS) for different values of K and identifies the point where additional clusters provide diminishing improvements. This point, known as the "elbow," represents the most suitable number of clusters. Choosing an appropriate number of clusters improves segmentation quality and ensures meaningful grouping of observations.

```python
# ============================================================
# ELBOW METHOD
# ============================================================

from sklearn.cluster import KMeans

X = scaled_data

wcss = []

for k in range(1, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X)

    wcss.append(
        kmeans.inertia_
    )

# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(10,6))

plt.plot(
    range(1,11),
    wcss,
    marker="o",
    markersize=8,
    linewidth=3,
    color="#6D597A"
)

plt.fill_between(
    range(1,11),
    wcss,
    alpha=0.2,
    color="#B56576"
)

plt.title(
    "Elbow Method for Optimal Clusters",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Number of Clusters (K)",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "WCSS",
    fontsize=12,
    fontweight="bold"
)

plt.grid(
    alpha=0.3,
    linestyle="--"
)

plt.tight_layout()

plt.show()

print("KEY INSIGHTS")
print("=" * 70)

print(
    f"• Maximum Clusters Tested: 10"
)

print(
    f"• Initial WCSS: {wcss[0]:.2f}"
)

print(
    f"• Final WCSS: {wcss[-1]:.2f}"
)

print(
    f"• Overall Reduction in WCSS: {reduction:.2f}%"
)

print(
    "• The elbow point indicates the optimal number of clusters."
)

print(
    "• Beyond the elbow, additional clusters provide limited improvement."
)
```

💡 Interpretation

</h2>

• The elbow curve illustrates how clustering quality improves as additional clusters are introduced.

• A steep decline in WCSS indicates substantial improvements in cluster compactness.

• The point where the curve begins to flatten represents the optimal number of clusters because additional clusters yield diminishing returns.

• Selecting too few clusters may oversimplify the dataset, while too many clusters can lead to over-segmentation.

• The elbow point provides a balance between clustering accuracy and model simplicity.

• The identified cluster count will be used in the next step to segment observations into meaningful groups based on their characteristics.

🚨 Anomaly Detection

</h2>

Anomaly Detection identifies observations that significantly differ from normal patterns in the dataset. Isolation Forest is an unsupervised machine learning algorithm specifically designed to detect unusual observations by isolating them from the rest of the data. These anomalies may represent exceptional business performance, operational inefficiencies, unusual customer behaviour, or potential data quality issues. Detecting anomalies helps improve data understanding and supports more accurate business decision-making.

```python
# ============================================================
# ANOMALY DETECTION + DONUT CHART
# ============================================================

from sklearn.ensemble import IsolationForest

anomaly_features = df[numerical_columns].copy()

anomaly_features = anomaly_features.fillna(
    anomaly_features.mean()
)

iso_forest = IsolationForest(
    contamination=0.05,
    random_state=42
)

df["Anomaly"] = iso_forest.fit_predict(
    anomaly_features
)

normal_count = (
    df["Anomaly"] == 1
).sum()

anomaly_count = (
    df["Anomaly"] == -1
).sum()

colors = [
    "#84A59D",  # Sage Green
    "#E56B6F"   # Coral
]

plt.figure(figsize=(8,8))

wedges, texts, autotexts = plt.pie(
    [normal_count, anomaly_count],
    labels=["Normal", "Anomaly"],
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={
        "width":0.45,
        "edgecolor":"white"
    },
    textprops={
        "fontsize":11,
        "fontweight":"bold"
    }
)

plt.text(
    0,
    0,
    f"{len(df):,}\nCompanies",
    ha="center",
    va="center",
    fontsize=14,
    fontweight="bold"
)

plt.title(
    "Anomaly Detection Distribution",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

print("KEY INSIGHTS")
print("=" * 70)

print(
    f"• Total Companies Analysed: {len(df):,}"
)

print(
    f"• Normal Companies: {normal_count:,}"
)

print(
    f"• Anomalous Companies: {anomaly_count:,}"
)

print(
    f"• Normal Share: {(normal_count/len(df))*100:.2f}%"
)

print(
    f"• Anomaly Share: {(anomaly_count/len(df))*100:.2f}%"
)
```

🤖 Train-Test Split

</h2>

Train-test splitting divides the dataset into two subsets: a training set used to build machine learning models and a testing set used to evaluate model performance on unseen data. This process helps assess model generalization and prevents overly optimistic performance estimates. A typical 80:20 split provides sufficient data for both training and validation.

```python
# ============================================================
# TRAIN TEST SPLIT
# ============================================================

from sklearn.model_selection import train_test_split

target_column = "market_rating"

X = df[numerical_columns].drop(
    columns=[target_column]
)

y = df[target_column]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# ============================================================
# DONUT CHART
# ============================================================

sizes = [
    len(X_train),
    len(X_test)
]

labels = [
    "Training Data",
    "Testing Data"
]

colors = [
    "#A8DADC",
    "#457B9D"
]

plt.figure(figsize=(8,8))

plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={
        "width":0.45,
        "edgecolor":"white"
    },
    textprops={
        "fontsize":11,
        "fontweight":"bold"
    }
)

plt.text(
    0,
    0,
    f"{len(df):,}\nRecords",
    ha="center",
    va="center",
    fontsize=14,
    fontweight="bold"
)

plt.title(
    "Train-Test Split",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()
plt.show()

print("KEY INSIGHTS")
print("=" * 70)

print(
    f"• Total Records: {len(df):,}"
)

print(
    f"• Training Records: {len(X_train):,}"
)

print(
    f"• Testing Records: {len(X_test):,}"
)

print(
    f"• Training Share: {(len(X_train)/len(df))*100:.2f}%"
)

print(
    f"• Testing Share: {(len(X_test)/len(df))*100:.2f}%"
)
```

🤖 Machine Learning Model Building

</h2>

Machine learning model building is used to predict the target variable using selected input features. In this project, multiple regression models are trained and evaluated to compare predictive performance. The models are assessed using R² Score, RMSE, and MAE, which help measure accuracy, error magnitude, and overall model reliability.

```python
# ============================================================
# MACHINE LEARNING MODEL BUILDING
# ============================================================

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),
    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    )
}

results = []

for model_name, model in models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    results.append({
        "Model": model_name,
        "R2 Score": r2_score(y_test, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_test, y_pred)),
        "MAE": mean_absolute_error(y_test, y_pred)
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
)

display(results_df)

best_model = results_df.iloc[0]

print("KEY INSIGHTS")
print("=" * 70)

print(f"• Best Performing Model: {best_model['Model']}")

print(f"• Best R² Score: {best_model['R2 Score']:.4f}")

print(f"• RMSE: {best_model['RMSE']:.4f}")

print(f"• MAE: {best_model['MAE']:.4f}")

print(f"• Total Models Compared: {len(results_df)}")
```

```python
# ============================================================
# RESIDUAL ANALYSIS
# ============================================================

residuals = y_test - y_pred

plt.figure(figsize=(9,6))

plt.scatter(
    y_pred,
    residuals,
    color="#B5838D",
    alpha=0.65,
    s=55,
    edgecolor="black",
    linewidth=0.3
)

plt.axhline(
    0,
    color="#6D6875",
    linestyle="--",
    linewidth=3
)

plt.title(
    "Residual Analysis",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Predicted Values",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    "Residuals",
    fontsize=12,
    fontweight="bold"
)

plt.grid(
    alpha=0.3,
    linestyle="--"
)

plt.tight_layout()
plt.show()

print("KEY INSIGHTS")
print("=" * 70)

print(f"• Mean Residual: {residuals.mean():.4f}")

print(f"• Minimum Residual: {residuals.min():.4f}")

print(f"• Maximum Residual: {residuals.max():.4f}")

print(f"• Residual Standard Deviation: {residuals.std():.4f}")

print(f"• Total Predictions Analysed: {len(residuals):,}")
```

📈 Time Trend Analysis

</h2>

Time Trend Analysis examines how a key business metric changes over time. By aggregating observations across dates and applying a rolling average, underlying patterns and long-term movements become easier to identify. This analysis helps detect growth trends, declines, fluctuations, and periods of stability within the dataset.

```python
# ============================================================
# MONTHLY TREND ANALYSIS
# ============================================================

date_column = "created_date"
target_column = "market_rating"

monthly_df = df.copy()

monthly_df[date_column] = pd.to_datetime(
    monthly_df[date_column]
)

monthly_trend = (
    monthly_df
    .groupby(
        monthly_df[date_column].dt.to_period("M")
    )[target_column]
    .mean()
    .reset_index()
)

monthly_trend[date_column] = (
    monthly_trend[date_column]
    .astype(str)
)

plt.figure(figsize=(12,6))

plt.plot(
    monthly_trend[date_column],
    monthly_trend[target_column],
    marker="o",
    linewidth=3,
    color="#52796F"
)

plt.title(
    "Monthly Trend Analysis",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Month")
plt.ylabel(target_column)

plt.xticks(rotation=45)

plt.grid(
    alpha=0.3,
    linestyle="--"
)

plt.tight_layout()
plt.show()

highest_month = monthly_trend.loc[
    monthly_trend[target_column].idxmax()
]

lowest_month = monthly_trend.loc[
    monthly_trend[target_column].idxmin()
]

print("KEY INSIGHTS")
print("=" * 70)

print(
    f"• Highest Month: {highest_month[date_column]}"
)

print(
    f"• Highest Average Rating: {highest_month[target_column]:.2f}"
)

print(
    f"• Lowest Month: {lowest_month[date_column]}"
)

print(
    f"• Lowest Average Rating: {lowest_month[target_column]:.2f}"
)

print(
    f"• Monthly Periods Analysed: {len(monthly_trend)}"
)

print(
    f"• Overall Monthly Average: {monthly_trend[target_column].mean():.2f}"
)
```

• Monthly analysis reveals short-term business fluctuations and seasonal behaviour.

• Months with higher average ratings indicate stronger business performance and customer satisfaction.

• Large fluctuations indicate volatility and changing business dynamics.

📈 Forecasting Analysis

</h2>

Forecasting analysis examines historical trends to estimate future values of a target variable. By analysing past patterns and smoothing fluctuations through moving averages, forecasting helps identify expected future behaviour and long-term business trends. This analysis supports strategic planning, resource allocation, and proactive decision-making.

```python
# ============================================================
# FORECASTING ANALYSIS
# ============================================================

date_column = "created_date"
target_column = "market_rating"

forecast_df = df.copy()

forecast_df[date_column] = pd.to_datetime(
    forecast_df[date_column]
)

forecast_df = (
    forecast_df
    .sort_values(date_column)
)

trend = (
    forecast_df
    .groupby(date_column)[target_column]
    .mean()
    .reset_index()
)

trend["Rolling_Avg"] = (
    trend[target_column]
    .rolling(7)
    .mean()
)

# ============================================================
# SIMPLE FORECAST
# ============================================================

forecast_value = (
    trend["Rolling_Avg"]
    .dropna()
    .tail(7)
    .mean()
)

future_dates = pd.date_range(
    start=trend[date_column].max()
          + pd.Timedelta(days=1),
    periods=30
)

forecast_data = pd.DataFrame({

    date_column: future_dates,

    "Forecast":
    forecast_value

})

# ============================================================
# PLOT
# ============================================================

plt.figure(figsize=(12,6))

plt.plot(
    trend[date_column],
    trend[target_column],
    color="#84A59D",
    alpha=0.5,
    label="Historical Trend"
)

plt.plot(
    trend[date_column],
    trend["Rolling_Avg"],
    color="#E9C46A",
    linewidth=3,
    label="7-Day Rolling Average"
)

plt.plot(
    forecast_data[date_column],
    forecast_data["Forecast"],
    color="#6D597A",
    linewidth=3,
    linestyle="--",
    label="Forecast"
)

plt.title(
    f"Forecasting Analysis - {target_column}",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel(
    "Date",
    fontsize=12,
    fontweight="bold"
)

plt.ylabel(
    target_column,
    fontsize=12,
    fontweight="bold"
)

plt.legend()

plt.grid(
    alpha=0.3,
    linestyle="--"
)

plt.tight_layout()
plt.show()

print("KEY INSIGHTS")
print("="*70)

print(
    f"• Historical Records Analysed: {len(trend):,}"
)

print(
    f"• Forecast Horizon: {len(future_dates)} Days"
)

print(
    f"• Latest Actual Value: {trend[target_column].iloc[-1]:.2f}"
)

print(
    f"• Forecasted Value: {forecast_value:.2f}"
)

print(
    f"• Average Historical Value: {trend[target_column].mean():.2f}"
)
```
